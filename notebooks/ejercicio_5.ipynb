## **5.5 Verificación de los nuevos archivos**

Se aplicaron las mismas validaciones del Ejercicio 2 a los archivos de 2024. Primero, a nivel de archivos, se comprobó contra el servidor de la TLC que los 24 archivos publicados (12 meses para cada tipo de taxi) existen localmente, tienen el mismo tamaño que informa el servidor y se pueden leer como Parquet. Luego, a nivel de contenido, se verificó con DuckDB que cada archivo tenga registros y que más del 99.9 % de sus viajes corresponda al mes que indica su nombre. En total se incorporaron 41,169,720 viajes de taxis amarillos y 660,218 de taxis verdes.

````python
import sys

import duckdb
import pandas as pd
import pyarrow.parquet as pq

sys.path.insert(0, str(Path("../scripts").resolve()))
import download_data as dd

ANIO_NUEVO = 2024
filas = []
for tipo in dd.TIPOS_TAXI:
    for mes in range(1, 13):
        estado, tamanio_servidor, _ = dd.consultar_publicacion(dd.construir_url(tipo, ANIO_NUEVO, mes))
        ruta = dd.ruta_destino(tipo, ANIO_NUEVO, mes)
        existe = ruta.exists()
        filas.append({
            "tipo": tipo,
            "mes": f"{ANIO_NUEVO}-{mes:02d}",
            "estado_servidor": estado,
            "existe_local": existe,
            "tamanio_local": ruta.stat().st_size if existe else None,
            "tamanio_servidor": tamanio_servidor,
            "filas": pq.ParquetFile(ruta).metadata.num_rows if existe else None,
        })

verificacion = pd.DataFrame(filas).astype(
    {"tamanio_local": "Int64", "tamanio_servidor": "Int64", "filas": "Int64"}
)
verificacion["coincide"] = (verificacion["tamanio_local"] == verificacion["tamanio_servidor"]).astype("boolean")

assert (verificacion["estado_servidor"] == dd.PUBLICADO).all(), "No todos los meses de 2024 están publicados"
assert verificacion["existe_local"].all(), "Falta descargar algún mes de 2024"
assert verificacion["coincide"].all(), "Algún archivo no coincide con el tamaño del servidor"
print(f"[ok] {len(verificacion)} archivos de {ANIO_NUEVO} descargados y con el tamaño del servidor")
verificacion
````

````python
con = duckdb.connect()

cobertura = con.sql(r"""
    WITH viajes AS (
        SELECT regexp_extract(filename, '(yellow|green)_tripdata', 1)        AS tipo,
               regexp_extract(filename, '_(\d{4}-\d{2})\.parquet', 1)      AS mes_archivo,
               strftime(COALESCE(tpep_pickup_datetime, lpep_pickup_datetime), '%Y-%m') AS mes_viaje
        FROM read_parquet('../data/raw/*/2024/*.parquet', filename = true, union_by_name = true)
    )
    SELECT tipo,
           mes_archivo,
           COUNT(*)                                                                      AS viajes,
           ROUND(100.0 * COUNT(*) FILTER (WHERE mes_viaje <> mes_archivo) / COUNT(*), 4) AS pct_fuera_del_mes
    FROM viajes
    GROUP BY tipo, mes_archivo
    ORDER BY tipo DESC, mes_archivo
""").df()

assert (cobertura["viajes"] > 0).all(), "Hay archivos sin registros"
assert (cobertura["pct_fuera_del_mes"] < 0.1).all(), "Algún archivo tiene muchos viajes fuera de su mes"
print(f"[ok] {len(cobertura)} archivos con registros y viajes dentro de su mes")
print(cobertura.groupby("tipo", sort=False)["viajes"].sum().to_string())
cobertura
````

## **5.6 Consulta conjunta de 2024 y 2026**

Con DuckDB es posible consultar ambos años en una sola consulta, ya que basta con que la ruta utilice un comodín en la carpeta del año (`data/raw/*/*/*.parquet`). La siguiente consulta obtiene, para cada año y tipo de taxi, la cantidad de viajes, los meses disponibles, la distancia mediana y el monto total mediano. Por ejemplo, se observa que en 2026 el monto mediano de un viaje amarillo subió de 21.00 a 23.58 dólares respecto a 2024, y que los taxis verdes pasaron de alrededor de 55 mil a 42 mil viajes por mes.

````python
RUTA_TODOS = "../data/raw/*/*/*.parquet"

con.sql(f"""
    SELECT regexp_extract(filename, '_(\\d{{4}})-', 1)                      AS anio,
           regexp_extract(filename, '(yellow|green)_tripdata', 1)            AS tipo,
           COUNT(*)                                                          AS viajes,
           COUNT(DISTINCT regexp_extract(filename, '_(\\d{{4}}-\\d{{2}})\\.parquet', 1)) AS meses,
           ROUND(COUNT(*) / COUNT(DISTINCT regexp_extract(filename, '_(\\d{{4}}-\\d{{2}})\\.parquet', 1))) AS viajes_por_mes,
           ROUND(MEDIAN(trip_distance), 2)                                   AS distancia_mediana,
           ROUND(MEDIAN(total_amount), 2)                                    AS total_mediano
    FROM read_parquet('{RUTA_TODOS}', filename = true, union_by_name = true)
    GROUP BY anio, tipo
    ORDER BY anio, tipo DESC
""").df()
````

## **5.7 Ajustes necesarios en las consultas desarrolladas**

En general, las consultas desarrolladas no necesitan modificarse para trabajar con el nuevo conjunto de archivos, ya que leen los datos con comodines en las rutas y con `union_by_name`, de tal forma que los archivos de 2024 se incorporan automáticamente. Por ejemplo, las vistas `viajes` y `viajes_limpios` del Ejercicio 4 funcionan sin cambios sobre ambos años, y la regla de limpieza de fechas compara contra el mes del archivo y no contra un año fijo. Sin embargo, al incorporar los nuevos archivos se identificaron los siguientes ajustes:

- **Esquema que cambia con el tiempo:** `cbd_congestion_fee` no existe en 2024 (el cobro inició en 2025) y `request_source` solo existe a partir de junio de 2026. Sin `union_by_name`, DuckDB toma únicamente el esquema del primer archivo e ignora estas columnas, por lo que se corrigieron los `DESCRIBE` de los incisos 3.3 y 3.4 para utilizarlo.
- **Año fijo en algunas reglas:** la regla `year(pickup) <> 2026` del inciso 3.6 y la verificación del Ejercicio 2 tienen el año escrito en el código, por lo que para otros años se debe utilizar la regla basada en el mes del archivo.
- **Agregaciones que mezclan años:** consultas como las de día de la semana, hora o percentiles agrupan todos los registros, por lo que al tener varios años se debe agregar el año al `GROUP BY` para no mezclar periodos.
- **Notebooks del análisis de 2026:** los Ejercicios 3 y 4 documentan resultados de 2026, por lo que se fijó su ruta a `data/raw/*/2026/*.parquet` mediante una constante `ANIO`. De esta forma, si se vuelven a ejecutar, siguen produciendo los mismos resultados, que se verificó ejecutándolos nuevamente después de descargar 2024.

````python
# Columnas que no existen en todos los archivos de un mismo tipo de taxi
con.sql(f"""
    WITH esquema AS (
        SELECT regexp_extract(file_name, '(yellow|green)_tripdata', 1) AS tipo,
               regexp_extract(file_name, '_(\\d{{4}})-', 1)           AS anio,
               file_name,
               name                                                   AS columna
        FROM parquet_schema('{RUTA_TODOS}')
        WHERE name <> 'schema'
    ),
    archivos_por_tipo AS (
        SELECT tipo, COUNT(DISTINCT file_name) AS archivos_totales
        FROM esquema
        GROUP BY tipo
    )
    SELECT e.tipo,
           e.columna,
           string_agg(DISTINCT e.anio, ', ' ORDER BY e.anio)  AS anios_con_la_columna,
           COUNT(DISTINCT e.file_name)                        AS archivos_con_la_columna,
           a.archivos_totales
    FROM esquema e
    JOIN archivos_por_tipo a USING (tipo)
    GROUP BY e.tipo, e.columna, a.archivos_totales
    HAVING COUNT(DISTINCT e.file_name) < a.archivos_totales
    ORDER BY e.tipo DESC, e.columna
""").df()
````

````python
# Las vistas del Ejercicio 4, sin cambios, sobre todos los anios descargados
con.sql(f"""
    CREATE OR REPLACE VIEW viajes AS
    SELECT regexp_extract(filename, '(yellow|green)_tripdata', 1)         AS tipo,
           regexp_extract(filename, '_(\\d{{4}}-\\d{{2}})\\.parquet', 1) AS mes_archivo,
           COALESCE(tpep_pickup_datetime, lpep_pickup_datetime)            AS pickup,
           COALESCE(tpep_dropoff_datetime, lpep_dropoff_datetime)          AS dropoff,
           passenger_count IS NULL AND RatecodeID IS NULL                  AS sin_metadatos,
           *
    FROM read_parquet('{RUTA_TODOS}', filename = true, union_by_name = true)
""")
con.sql("""
    CREATE OR REPLACE VIEW viajes_limpios AS
    SELECT *,
           epoch(dropoff - pickup) / 60 AS duracion_min
    FROM viajes
    WHERE strftime(pickup, '%Y-%m') = mes_archivo
      AND dropoff > pickup AND dropoff - pickup <= INTERVAL 24 HOUR
      AND trip_distance > 0 AND trip_distance <= 100
      AND fare_amount >= 0 AND total_amount > 0
""")

con.sql("""
    SELECT year(pickup)                                            AS anio,
           COUNT(*)                                                AS viajes_limpios,
           ROUND(100 * AVG(sin_metadatos::INT), 1)                 AS pct_sin_metadatos,
           ROUND(MEDIAN(trip_distance), 2)                         AS distancia_mediana,
           ROUND(MEDIAN(duracion_min), 1)                          AS duracion_mediana_min
    FROM viajes_limpios
    GROUP BY anio
    ORDER BY anio
""").df()
````

## **5.9 Características del diseño que permiten incorporar nuevos archivos**

Considero que el sistema permite incorporar nuevos archivos sin modificar todo el flujo de análisis gracias a varias decisiones de diseño. Primero, el script de descarga recibe los años como parámetro (`--anio` y `ANIOS_POR_DEFECTO`) en lugar de tenerlos fijos en el código, y omite los archivos que ya existen, por lo que incorporar un año nuevo es cuestión de agregarlo y volver a ejecutar el mismo script, el cual solo descarga lo que falta. Además, el script trabaja con rutas resueltas a partir de su propia ubicación y guarda los archivos en una estructura predecible (`data/raw/<tipo>/<anio>/`), de tal forma que el resultado es el mismo sin importar desde dónde se ejecute.

Por otro lado, las consultas leen los archivos con comodines en la ruta y con `union_by_name`, por lo que DuckDB incorpora automáticamente los archivos nuevos y absorbe las columnas que la TLC agrega o elimina con el tiempo; asimismo, la lógica de unificación y limpieza está centralizada en las vistas `viajes` y `viajes_limpios`, por lo que no es necesario repetirla en cada consulta. Finalmente, todo se ejecuta dentro de un ambiente reproducible con Docker, lo que garantiza que cualquier persona del equipo obtenga los mismos resultados con las mismas versiones de las herramientas.
